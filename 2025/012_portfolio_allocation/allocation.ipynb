# 🚀 Stratégie 1 : Volatility Targeting

## 💡 Concept pédagogique :
Le **Volatility Targeting** est l'une des approches les plus intuitives en allocation de portefeuille. L'idée est simple : **allouer plus de capital aux actifs stables et moins aux actifs volatils**.

### 🧠 Logique :
- Les actifs moins volatils sont généralement moins risqués
- On peut donc leur allouer une part plus importante du portefeuille
- L'allocation est **inversement proportionnelle** à la volatilité historique

### ✅ Avantages :
- **Simplicité** : facile à comprendre et implémenter
- **Stabilité** : tend à réduire la volatilité globale du portefeuille
- **Robustesse** : moins sensible aux erreurs d'estimation que d'autres méthodes

### ⚠️ Limites :
- Ignore complètement les **rendements espérés**
- Ne tient pas compte des **corrélations** entre actifs
- Peut surpondérer des actifs à faible rendement
